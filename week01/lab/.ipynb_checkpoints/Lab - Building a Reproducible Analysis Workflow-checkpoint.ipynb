## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

## Investigation 2

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 2?**

Your answer:Is the percentage of fatal accidents higher on weekends than on weekdays?


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer:I predict that the percentage of fatal accidents will be higher on weekends because there may be more impaired driving and increased travel during that time.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer:I will use each row as one accident and focus on Accident Date and Accident_Severity. I will group accidents by weekend and weekday, then calculate the percentage that were fatal out of all accidents in each group. I will exclude records with missing dates or severity values.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results

accidents_q2 = accidents_df.dropna(subset=['Accident Date', 'Accident_Severity']).copy()

accidents_q2['Accident Date'] = pd.to_datetime(
    accidents_q2['Accident Date'], dayfirst=True
)

accidents_q2['Day_Type'] = 'Weekday'

accidents_q2.loc[
    accidents_q2['Accident Date'].dt.dayofweek >= 5, 'Day_Type'
] = 'Weekend'
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
fatal_count = accidents_q2[accidents_q2['Accident_Severity'] == 'Fatal'].groupby('Day_Type').size()

total_count = accidents_q2['Day_Type'].value_counts()

fatal_percentage = fatal_count / total_count * 100

fatal_percentage
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:The weekend fatal accident percentage was 1.45%, compared to 1.26% on weekdays. I divided the fatal accidents by the total accidents in each group and multiplied by 100 to compare the percentages.


### Step 4: Check

Perform a sanity check tied to the meaning of your question. Explain what values or patterns would make you suspicious and whether your result makes sense. Depending on the analysis, consider plausible values and units, whether indicators match original labels, or whether the ranges within comparison groups match their definitions. Use additional code only when it helps this check; do not simply repeat the initial row count or missing-value inspection. An unexpected result is a reason to investigate, not to change the result to fit your prediction.


```python
# Optional: add a targeted sanity check if code would help.
accidents_q2.groupby('Day_Type')['Accident_Severity'].value_counts()
```

**What would make you suspicious, and why? Does your evidence make sense in context? Describe any issue you found and how you addressed it.**

Your answer:I would be suspicious if the fatal accident percentages were unusually high or if the severity labels were incorrect. My results make sense because both groups contain fatal accidents, and weekends have a slightly higher percentage than weekdays. I also made sure the dates were read in the correct day-month-year format.


### Step 5: Conclusion


**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**

Your answer:The percentage of fatal accidents was higher on weekends (1.45%) than weekdays (1.26%), which supports my prediction.


### Step 6: Limitation


**What should a reader avoid concluding from this result, and why? Connect your limitation to this question and dataset.**

Your answer:A reader should not conclude that weekends cause more fatal accidents because other factors, such as weather, traffic, and impaired driving, may also contribute to accident severity.

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 3?**

Your answer: Which days of the week has the highest number of accidents?


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer:I predict that Friday will have the highest number of accidents because traffic may increase as people travel for work, errands, and weekend activities.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer:I will use each row as one accident and the Accident Date feature to identify the day of the week. I will count the number of accidents for each day and exclude records with missing dates.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# Prepare, filter, or calculate features as needed; display useful intermediate results.
accidents_q3 = accidents_df.dropna(subset=['Accident Date']).copy()

accidents_q3['Accident Date'] = pd.to_datetime(
    accidents_q3['Accident Date'], dayfirst=True
)

accidents_q3['Day_of_Week'] = accidents_q3['Accident Date'].dt.day_name()
```

```python
# Calculate and display the evidence that answers your question. Add cells as needed.
accidents_q3['Day_of_Week'].value_counts()
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:Saturday had the highest number of accidents, with 108,607. I used the accident dates to identify each day of the week and counted how many accidents occurred on each day.

### Step 4: Check

Perform a sanity check tied to the meaning of your question. Explain what values or patterns would make you suspicious and whether your result makes sense. Depending on the analysis, consider plausible values and units, whether indicators match original labels, or whether the ranges within comparison groups match their definitions. Use additional code only when it helps this check; do not simply repeat the initial row count or missing-value inspection. An unexpected result is a reason to investigate, not to change the result to fit your prediction.


```python
# Optional: add a targeted sanity check if code would help.
print(accidents_q3['Day_of_Week'].value_counts().sum())

print(accidents_q3['Day_of_Week'].nunique())
```

**What would make you suspicious, and why? Does your evidence make sense in context? Describe any issue you found and how you addressed it.**

Your answer:I would be suspicious if any days were missing or had unusually low accident counts. My results make sense because all seven days are represented, and Saturday had the highest number of accidents. I excluded records with missing dates before counting.


### Step 5: Conclusion


**Answer your research question in a sentence using the numerical evidence. How does it compare with your prediction?**

Your answer:Saturday had the highest number of accidents, with 108,607. This did not support my prediction because I expected Friday to have the highest number.

**What should a reader avoid concluding from this result, and why? Connect your limitation to this question and dataset.**

Your answer:A reader should not conclude that Saturday causes more accidents because other factors, such as traffic volume, weather, and driving conditions, may influence accident numbers.
